```python
import pandas as pd

# === Configuration ===
NUM_ENTITIES = 100
NUM_ENTITY_TYPES = 10
NUM_ATTRIBUTES = 500
MIN_REL_PER_ENTITY = 1
MAX_REL_PER_ENTITY = 5

# Token IDs
entity_tokens = list(range(1000, 1000 + NUM_ENTITIES))
type_tokens = list(range(2000, 2000 + NUM_ENTITY_TYPES))
attribute_tokens = list(range(3000, 3000 + NUM_ATTRIBUTES))

# === Deterministic Assignment Functions ===

def deterministic_types_for_entity(entity_id):
    """Assigns 1-5 entity types based on entity_id (reproducible)"""
    num_types = (entity_id % (MAX_REL_PER_ENTITY - MIN_REL_PER_ENTITY + 1)) + MIN_REL_PER_ENTITY
    return [type_tokens[(entity_id + i) % NUM_ENTITY_TYPES] for i in range(num_types)]

def deterministic_attribute(entity_id, type_id):
    """Assigns a deterministic attribute for a (entity, type) pair"""
    index = (entity_id * 31 + type_id * 17) % NUM_ATTRIBUTES
    return attribute_tokens[index]

# === Dataset Creation ===
dataset = []

for entity_id in entity_tokens:
    type_ids = deterministic_types_for_entity(entity_id)
    for type_id in type_ids:
        attr_id = deterministic_attribute(entity_id, type_id)
        dataset.append((entity_id, type_id, attr_id))

# === Save Dataset ===
df = pd.DataFrame(dataset, columns=["entity", "entity_type", "attribute"])
dataset_path = "deterministic_entity_dataset.csv"
df.to_csv(dataset_path, index=False)
print(f"Saved deterministic dataset with {len(df)} rows to {dataset_path}")
```
